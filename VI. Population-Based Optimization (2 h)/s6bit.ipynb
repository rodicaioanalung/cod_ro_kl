# bitstring representation

In Julia, you can work with binary representations of floating-point numbers using functions like `bitstring`, `reinterpret`, and `parse`. Here's how you can convert a float to its binary representation and back:



## Convert Float to Binary Representation
To convert a float into its binary representation, you can use the `bitstring` function:
- `bitstring` gives a string representing the binary form of the floating-point number, including the sign, exponent, and mantissa.



```julia
x = 3.14  # Example float
binary_representation = bitstring(x)
println("Binary representation: $binary_representation")
```

## Convert Binary Representation Back to Float
To convert the binary representation back to a float:

Use `parse` to create an integer from the binary string.
Use `reinterpret` to interpret the integer as a float.



```julia
# Convert binary representation back to float
integer_representation = parse(Int64, binary_representation; base=2)
@show integer_representation
x_reconstructed = reinterpret(Float64, integer_representation)
println("Reconstructed float: $x_reconstructed")
```

- Ensure the bit length matches the floating-point type (e.g., Float32 for 32 bits or Float64 for 64 bits).
- These methods operate on the raw binary form, so the conversion captures the exact underlying representation.


## Binary Representation of Data
All data in memory is stored as a sequence of bits. For example:

A 64-bit Float64 (double-precision floating point) uses:  
- 1 bit for the sign.
- 11 bits for the exponent.
- 52 bits for the mantissa (fractional part).  

A 64-bit Int64 (signed integer) uses:  
- All 64 bits to represent the integer in two's complement format.


The same sequence of 64 bits can represent entirely different values depending on whether you interpret it as a Float64, Int64, or another type.

```julia
# Define a negative number
x = -3.14

# Convert to binary representation
binary_representation = bitstring(x)
println("Binary representation of -3.14: $binary_representation")

# Convert binary back to float
integer_representation = parse(UInt64, binary_representation; base=2)
reconstructed_float = reinterpret(Float64, integer_representation)
println("Reconstructed float: $reconstructed_float")
```
