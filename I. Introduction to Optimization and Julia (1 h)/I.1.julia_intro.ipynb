## 5. Vectors

Vectors are one-dimensional arrays. Julia uses **1-based indexing**.


````julia
v1 = zeros(3)
v2 = ones(3)
v3 = rand(3)
v4 = [3, 1, 4]
v5 = [3.1415, 1.618, 2.7182]

println(v4)
println(v5[1])
println(v5[end])
println(v5[end-1])
````

## 7. Copying versus Reference

When assigning a vector to another variable, both variables may refer to the same object. Use `copy` to create an independent copy.


````julia
x = [1, 2, 3]
y = x
y[1] = 4
println(x)

y = copy(x)
y[1] = 100
println("x = ", x)
println("y = ", y)
````

## 10. Linear Systems and Least Squares

The `\` operator efficiently solves linear systems and least-squares problems.


````julia
A = [1 0; 1 -2]
b = [32, -4]
x = A \ b
println(x)
println(A*x - b)

A2 = [1 2; 3 4; 5 6]
b2 = [0, 8, 9]
x2 = A2 \ b2
println(x2)
println(norm(A2*x2 - b2))
````

````julia
function f(x, y)
    return x + y
end

f_compact(x, y) = x + y
h = x -> x^2 + 1

g(fun, a, b) = [fun(a), fun(b)]

println(f(3, 4))
println(f_compact(3, 0.1415))
println(h(3))
println(g(h, 5, 10))
````

## 18. The `if` Statement


````julia
function compare_xy(x, y)
    if x < y
        return "x < y"
    elseif x > y
        return "x > y"
    else
        return "x = y"
    end
end

println(compare_xy(2, 5))
````
