# Plotting Functions in Julia

## One-variable and two-variable functions, points, and lines

This tutorial introduces plotting in Julia using `Plots.jl`. We will learn how to:

- plot a function of one variable \(f(x)\);
- customize axes, labels, legends, and line styles;
- add individual points and sets of points to a plot;
- add line segments and reference lines;
- plot several functions on the same figure;
- plot a function of two variables \(f(x,y)\) as a surface;
- visualize a two-variable function using contour and heatmap plots;
- add points and curves to 3D plots.

The notebook ends with exercises for students.


## 1. Setup

We use the `Plots.jl` package. If it is not installed, uncomment the installation line and run the cell once.


```julia
# using Pkg
# Pkg.add("Plots")

using Plots
```

## 2. Plotting a function of one variable

Consider

\[
f(x)=x^2-2x+1.
\]

In Julia, a function can be defined using the compact syntax below.


```julia
f(x) = x^2 - 2x + 1

plot(f, -3, 5,
    xlabel = "x",
    ylabel = "f(x)",
    title = "Plot of f(x) = x² - 2x + 1",
    label = "f(x)",
    linewidth = 2)
```

### Using an explicit grid of x-values

Sometimes we want to control exactly where the function is evaluated. We can create a range of values and use broadcasting (`f.(x)`) to evaluate the function at every point.


```julia
x = range(-3, 5, length=200)
y = f.(x)

plot(x, y,
    xlabel="x",
    ylabel="f(x)",
    label="f(x)",
    linewidth=2,
    title="Function evaluated on a grid")
```

## 3. Adding points to a plot

The mutating function `scatter!` adds points to an existing plot. The exclamation mark indicates that the current plot is modified.


```julia
p = plot(f, -3, 5,
    xlabel="x",
    ylabel="f(x)",
    label="f(x)",
    linewidth=2)

x_points = [0, 1, 2, 3]
y_points = f.(x_points)

scatter!(p, x_points, y_points,
    label="Selected points",
    markersize=6)

p
```

### Adding a single point

A single point can be highlighted in exactly the same way by passing one-element vectors.


```julia
p = plot(f, -3, 5, label="f(x)", linewidth=2)

x0 = 1.0
y0 = f(x0)

scatter!(p, [x0], [y0],
    label="Point (1, f(1))",
    markersize=8)

p
```

## 4. Adding lines and line segments

`plot!` can also add a line to an existing figure. To draw a segment between two points, provide their x- and y-coordinates.


```julia
p = plot(f, -3, 5,
    label="f(x)",
    linewidth=2,
    xlabel="x",
    ylabel="y")

x_line = [-1, 3]
y_line = [4, 4]

plot!(p, x_line, y_line,
    label="Line segment",
    linewidth=2,
    linestyle=:dash)

scatter!(p, x_line, y_line,
    label="Endpoints",
    markersize=5)

p
```

### Horizontal and vertical reference lines

`hline!` and `vline!` are convenient for adding reference lines.


```julia
p = plot(f, -3, 5, label="f(x)", linewidth=2)

hline!(p, [2], label="y = 2", linestyle=:dash)
vline!(p, [1], label="x = 1", linestyle=:dot)

p
```

## 5. Plotting several functions

Suppose we want to compare

\[
f(x)=x^2
\qquad\text{and}\qquad
g(x)=2x+1.
\]

We first create one plot and then add the second function using `plot!`.


```julia
f1(x) = x^2
g1(x) = 2x + 1

p = plot(f1, -3, 4,
    label="f(x) = x²",
    linewidth=2,
    xlabel="x",
    ylabel="y",
    title="Two functions")

plot!(p, g1, -3, 4,
    label="g(x) = 2x + 1",
    linewidth=2)

p
```

## 6. Example: function, point, and tangent line

This example combines the ideas above. For \(f(x)=x^2\), the derivative is \(f'(x)=2x\). At \(x_0=1\), the tangent has slope 2.


```julia
f2(x) = x^2
x0 = 1.0
y0 = f2(x0)
slope = 2*x0

tangent(x) = y0 + slope*(x - x0)

p = plot(f2, -2, 3,
    label="f(x) = x²",
    linewidth=2,
    xlabel="x",
    ylabel="y",
    title="Function and tangent line")

plot!(p, tangent, -2, 3,
    label="Tangent at x = 1",
    linestyle=:dash,
    linewidth=2)

scatter!(p, [x0], [y0],
    label="Tangency point",
    markersize=7)

p
```

# Part II — Functions of Two Variables

## 7. Defining a function \(f(x,y)\)

Consider

\[
f(x,y)=x^2+y^2.
\]

A function of two variables can be represented as a three-dimensional surface, where the height \(z\) is equal to \(f(x,y)\).


```julia
f(x, y) = x^2 + y^2

x = range(-3, 3, length=80)
y = range(-3, 3, length=80)

surface(x, y, f,
    xlabel="x",
    ylabel="y",
    zlabel="f(x,y)",
    title="Surface: f(x,y) = x² + y²",
    label="")
```

## 8. Another two-variable function

Surfaces are especially useful for understanding peaks, valleys, and oscillations.


```julia
g(x, y) = sin(x) * cos(y)

x = range(-2*pi, 2*pi, length=100)
y = range(-2*pi, 2*pi, length=100)

surface(x, y, g,
    xlabel="x",
    ylabel="y",
    zlabel="g(x,y)",
    title="Surface: sin(x) cos(y)",
    label="")
```

## 9. Contour plots

A contour plot represents a surface using curves of equal function value. It is often easier to read than a 3D surface when we are interested in level sets.


```julia
f(x, y) = x^2 + y^2

x = range(-3, 3, length=100)
y = range(-3, 3, length=100)

contour(x, y, f,
    xlabel="x",
    ylabel="y",
    title="Contour plot of x² + y²",
    fill=false)
```

### Filled contours


```julia
contourf(x, y, f,
    xlabel="x",
    ylabel="y",
    title="Filled contour plot of x² + y²")
```

## 10. Heatmaps

For a heatmap, it is useful to explicitly build the matrix of function values. Each entry `Z[j,i]` corresponds to the value at `(x[i], y[j])`.


```julia
x = range(-3, 3, length=100)
y = range(-3, 3, length=100)

Z = [f(xi, yi) for yi in y, xi in x]

heatmap(x, y, Z,
    xlabel="x",
    ylabel="y",
    title="Heatmap of f(x,y) = x² + y²")
```

## 11. Adding points to a 3D surface

For a point \((x_0,y_0)\), its height on the surface is \(z_0=f(x_0,y_0)\). We can add it using `scatter3d!`.


```julia
f(x, y) = x^2 + y^2

x = range(-3, 3, length=80)
y = range(-3, 3, length=80)

p = surface(x, y, f,
    xlabel="x",
    ylabel="y",
    zlabel="z",
    title="Surface with selected points",
    label="")

xp = [1.0, -1.5, 2.0]
yp = [1.0,  0.5, -1.0]
zp = f.(xp, yp)

scatter3d!(p, xp, yp, zp,
    label="Points on surface",
    markersize=5)

p
```

## 12. Adding a curve to a 3D surface

A line in the \(xy\)-plane can be lifted onto the surface. For example, along \(y=x\), the curve on the surface is

\[
(x,\;x,\;f(x,x)).
\]


```julia
f(x, y) = x^2 + y^2

x = range(-3, 3, length=80)
y = range(-3, 3, length=80)

p = surface(x, y, f,
    xlabel="x",
    ylabel="y",
    zlabel="z",
    title="Surface with the curve y = x",
    label="")

t = range(-3, 3, length=100)
z_curve = f.(t, t)

plot3d!(p, t, t, z_curve,
    label="Curve y = x",
    linewidth=3)

p
```

## 13. Adding points to a contour plot

For a contour plot, only the \(x\)- and \(y\)-coordinates are needed because the contours already encode the function value.


```julia
f(x, y) = x^2 + y^2

x = range(-3, 3, length=100)
y = range(-3, 3, length=100)

p = contour(x, y, f,
    xlabel="x",
    ylabel="y",
    title="Contour plot with points")

scatter!(p, [0, 1, -2], [0, 1, 1],
    label="Selected points",
    markersize=6)

p
```

# Summary

The most useful `Plots.jl` commands introduced in this notebook are:

| Task | Command |
|---|---|
| Plot a 1D function | `plot(f, a, b)` |
| Add another curve | `plot!` |
| Add points | `scatter!` |
| Add horizontal lines | `hline!` |
| Add vertical lines | `vline!` |
| Plot a surface | `surface` |
| Plot contours | `contour`, `contourf` |
| Plot a heatmap | `heatmap` |
| Add 3D points | `scatter3d!` |
| Add a 3D curve | `plot3d!` |

The `!` versions modify an existing plot. This makes it easy to build a visualization step by step by combining functions, points, and lines.
